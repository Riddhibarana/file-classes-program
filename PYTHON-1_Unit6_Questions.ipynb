## Question 11

Write a “pager” program. Your solution should prompt for a filename, and display the text file 25 lines at a time, pausing each time to ask the user to enter the word “continue”, in order to show the next 25 lines or enter the word “stop” to close the file.

*Difficulty: Medium*

```python
f = open("mbox-short.txt", "r")
x = f.readlines()

i = 0

while i < len(x):
    for j in range(i, i+25):
        if j < len(x):
            print(x[j])

    i = i + 25

    if i < len(x):
        a = input("Enter continue or stop: ")

        if a.lower() == "stop":
            break
        elif a.lower() == "continue":
            continue

f.close()
```

## Question 15

Write a Python program to count words, characters and spaces from text file.

Input:

Python is a Easy Subject
OOPs is One of the most
interesting Topic

Output:

No of space: 10
No of word: 13
No of character: 64

*Difficulty: Medium*

```python
f = open("st.txt", "r")
x = f.readlines()
f.close()

w = 0
s = 0
c = 0

for i in x:
    w = w + len(i.split())
    s = s + i.count(" ")
    c = c + len(i.strip())

print("No of space:", s)
print("No of word:", w)
print("No of character:", c)
```

## Question 17

Write a python program to accept string/sentence from user till the user enters “END”. Each string/sentence entered by user should be a newline in file. Save all the lines in file and display only those lines which begin with capital letter.

Example:
Enter Something (for quit enter END):Hi Friends
Enter Something (for quit enter END):how are you all
Enter Something (for quit enter END):I am fine
Enter Something (for quit enter END):hope you all are fine
Enter Something (for quit enter END):END
The Line started with Capital Letters:
Hi Friends
I am fine

*Difficulty: Medium*

```python
f = open("h.txt", "w")

while True:
    x = input("Enter Something (for quit enter END):")

    if x == "END" or x=="end":
        break

    f.write(x + "\n")

f.close()

f = open("h.txt", "r")
y = f.readlines()
f.close()

print("The Line started with Capital Letters:")

for i in y:
    if i[0].isupper():
        print(i, end="")
```
